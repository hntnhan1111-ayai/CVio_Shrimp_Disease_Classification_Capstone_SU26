## 3. Add ECAAttention and SimAM into `conv.py`

````python
from pathlib import Path

conv_path = ULTRA_REPO / "ultralytics" / "nn" / "modules" / "conv.py"
text = conv_path.read_text(encoding="utf-8")

eca_block = """
class ECAAttention(nn.Module):
    # ECA-Net: Efficient Channel Attention.
    # c1 is kept for Ultralytics YAML parser compatibility.
    def __init__(self, c1=None, k_size=3):
        super().__init__()
        self.avg_pool = nn.AdaptiveAvgPool2d(1)
        self.conv = nn.Conv1d(
            1,
            1,
            kernel_size=k_size,
            padding=(k_size - 1) // 2,
            bias=False
        )
        self.sigmoid = nn.Sigmoid()

    def forward(self, x):
        # x: [B, C, H, W]
        y = self.avg_pool(x)                       # [B, C, 1, 1]
        y = y.squeeze(-1).transpose(-1, -2)        # [B, 1, C]
        y = self.conv(y)                           # [B, 1, C]
        y = self.sigmoid(y).transpose(-1, -2).unsqueeze(-1)  # [B, C, 1, 1]
        return x * y.expand_as(x)
"""

simam_block = """
class SimAM(nn.Module):
    # SimAM: Simple, Parameter-Free Attention Module.
    # c1 is kept for Ultralytics YAML parser compatibility.
    def __init__(self, c1=None, e_lambda=1e-4):
        super().__init__()
        self.e_lambda = e_lambda
        self.activation = nn.Sigmoid()

    def forward(self, x):
        b, c, h, w = x.size()
        n = h * w - 1
        if n <= 0:
            return x

        x_minus_mu_square = (x - x.mean(dim=[2, 3], keepdim=True)).pow(2)
        y = x_minus_mu_square / (
            4 * (x_minus_mu_square.sum(dim=[2, 3], keepdim=True) / n + self.e_lambda)
        ) + 0.5

        return x * self.activation(y)
"""

updated = text
if "class ECAAttention(nn.Module):" not in updated:
    updated += "\n\n" + eca_block
if "class SimAM(nn.Module):" not in updated:
    updated += "\n\n" + simam_block

conv_path.write_text(updated, encoding="utf-8")
print("Updated:", conv_path)
````

## 4. Export ECAAttention and SimAM in `__init__.py`

````python
init_path = ULTRA_REPO / "ultralytics" / "nn" / "modules" / "__init__.py"
text = init_path.read_text(encoding="utf-8")

if "ECAAttention" not in text:
    if "from .conv import (" in text:
        text = text.replace(
            "from .conv import (",
            "from .conv import (\n    ECAAttention,\n    SimAM,"
        )
    else:
        text += "\nfrom .conv import ECAAttention, SimAM\n"

if "__all__" in text and "ECAAttention" not in text.split("__all__", 1)[1]:
    text = text.replace(
        '"Concat",',
        '"Concat",\n    "ECAAttention",\n    "SimAM",'
    )

init_path.write_text(text, encoding="utf-8")
print("Updated:", init_path)
````

## 5. Import/register ECAAttention and SimAM in `tasks.py`

````python
tasks_path = ULTRA_REPO / "ultralytics" / "nn" / "tasks.py"
text = tasks_path.read_text(encoding="utf-8")

if "ECAAttention" not in text:
    text = text.replace(
        "from ultralytics.nn.modules import (",
        "from ultralytics.nn.modules import (\n    ECAAttention,\n    SimAM,"
    )

anchor = "elif m in frozenset({Detect, WorldDetect, Segment, Pose, OBB, ImagePoolingAttn, v10Detect}):"
helper = """elif m is ECAAttention:
            c2 = ch[f]
            k_size = args[1] if len(args) > 1 else 3
            args = [c2, k_size]
        elif m is SimAM:
            c2 = ch[f]
            args = [c2, *args]
        """

if "elif m is ECAAttention:" not in text and anchor in text:
    text = text.replace(anchor, helper + anchor)

tasks_path.write_text(text, encoding="utf-8")
print("Updated:", tasks_path)
````

## 6. Create YAML model: `yolo11n-seg-eca-simam-head.yaml`

This YAML inserts **ECA first**, then **SimAM**, then feeds the refined P3/P4/P5 feature maps into the `Segment` head.

Design:

```text
P3 -> ECA -> SimAM
P4 -> ECA -> SimAM
P5 -> ECA -> SimAM
[refined P3, refined P4, refined P5] -> Segment
```


````python
cfg_dir = ULTRA_REPO / "ultralytics" / "cfg" / "models" / "11"
base_yaml = cfg_dir / "yolo11-seg.yaml"
new_yaml = cfg_dir / "yolo11n-seg-eca-simam-head.yaml"

text = base_yaml.read_text(encoding="utf-8")

old = "- [[16, 19, 22], 1, Segment, [nc, 32, 256]]"
new = """# ECA + SimAM inserted before Segment head for P3, P4, P5 features
  - [16, 1, ECAAttention, [256, 3]]
  - [23, 1, SimAM, []]
  - [19, 1, ECAAttention, [512, 3]]
  - [25, 1, SimAM, []]
  - [22, 1, ECAAttention, [1024, 3]]
  - [27, 1, SimAM, []]
  - [[24, 26, 28], 1, Segment, [nc, 32, 256]]"""

if old in text:
    text = text.replace(old, new)
else:
    print("WARNING: Expected Segment line not found. Please manually inspect base YAML.")

new_yaml.write_text(text, encoding="utf-8")
print("Created:", new_yaml)
print(new_yaml.read_text(encoding="utf-8")[-1200:])
````

## 15. Validate best checkpoint

````python
BEST_PT = Path(PROJECT_DIR) / EXP_NAME / "weights" / "best.pt"
print("BEST_PT:", BEST_PT)

best_model = YOLO(str(BEST_PT))

metrics = best_model.val(
    data=DATA_YAML,
    imgsz=IMG_SIZE,
    device=DEVICE,
    split="val"
)

print("Box mAP50:", metrics.box.map50)
print("Box mAP50-95:", metrics.box.map)

if hasattr(metrics, "seg"):
    print("Mask mAP50:", metrics.seg.map50)
    print("Mask mAP50-95:", metrics.seg.map)
else:
    print("No segmentation metrics found.")
````
