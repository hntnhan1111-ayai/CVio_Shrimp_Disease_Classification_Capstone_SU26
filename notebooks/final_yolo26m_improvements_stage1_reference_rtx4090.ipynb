# Final YOLOv26m Improvement Runs Against Fixed Stage 1 Baseline

This notebook does **not** rerun the CE baseline. It uses the corrected Stage 1 native Ultralytics YOLOv26m-cls baseline as the fixed reference:

- Test Macro-F1: 0.8902
- Accuracy: 0.8902
- Cohen’s Kappa: 0.8505

It runs only improvement variants: ASLSingleLabel, 10 ASL-derived custom losses, 8 added attention modules, and the best-loss + best-attention combined variant.

```python
from pathlib import Path
import sys, os, subprocess, json

NOTEBOOKS_ROOT = Path(os.environ.get("PROJECT_ROOT", Path.cwd())).resolve()
RUNNER = NOTEBOOKS_ROOT / "final_yolo26m_improvements_stage1_reference_runner.py"
PATCHER = NOTEBOOKS_ROOT / "patch_yolo_attention_injection_v2.py"
PROJECT_DIR = NOTEBOOKS_ROOT / "Improving Lightweight Shrimp Disease Classification with Co-Infection-Aware Losses and RandAugment"
OUTPUT_DIR = NOTEBOOKS_ROOT / "final_yolo26m_improvements_stage1_reference_outputs"

print("Python:", sys.executable)
print("Project:", PROJECT_DIR, PROJECT_DIR.exists())
print("Runner:", RUNNER, RUNNER.exists())
print("Patcher:", PATCHER, PATCHER.exists())
print("Output:", OUTPUT_DIR)
assert RUNNER.exists(), f"Missing runner script: {RUNNER}"
assert PATCHER.exists(), f"Missing patch script: {PATCHER}"
assert PROJECT_DIR.exists(), f"Missing project directory: {PROJECT_DIR}"
```

```python
cmd = [
    sys.executable, str(RUNNER),
    "--project_dir", str(PROJECT_DIR),
    "--output_dir", str(OUTPUT_DIR),
    "--patch_script", str(PATCHER),
    "--epochs", "30",
]
print("RUN:", " ".join(cmd))
subprocess.run(cmd, check=True)
```
